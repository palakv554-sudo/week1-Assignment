**Q10**. Why is NumPy significantly faster than Python lists for numerical operations? Give 3 reasons, at least one should mention memory layout.

A10.
1. **Memory layout:** an `ndarray` is one contiguous block of same-sized raw numbers; a list is an array of pointers to separate Python objects scattered in memory, so the CPU cache works well for arrays and badly for lists.

**Q31**. Find all Armstrong numbers between 1 and 9999. (An n-digit number is Armstrong if the sum of its digits each raised to the power n equals the number itself. Example: 153 = 1³ + 5³ + 3³.) Print them in a clean comma-separated list with a count at the end.

````python
armstrong = []
for n in range(1, 10000):
    digits, temp = 0, n
    while temp > 0:               # count digits
        digits += 1
        temp //= 10
    total, temp = 0, n
    while temp > 0:               # sum of digit ** digits
        total += (temp % 10) ** digits
        temp //= 10
    if total == n:
        armstrong.append(n)

for idx in range(len(armstrong)):
    print(armstrong[idx], end=", " if idx < len(armstrong) - 1 else "\n")
print("Count:", len(armstrong))
````

**Q33**. Prime check, optimized. Take a number N. Check if it is prime by iterating only up to √N. Then, print the SUM of all primes between 2 and N (inclusive). Do not use any library.

````python
n = int(input("Enter N: "))

# Part 1: Check if N itself is prime
if n < 2:
    print(f"{n} is NOT a prime number.")
else:
    is_n_prime = True
    limit = int(n ** 0.5)
    for d in range(2, limit + 1):
        if n % d == 0:
            is_n_prime = False
            break

    if is_n_prime:
        print(f"{n} is a PRIME number.")
    else:
        print(f"{n} is NOT a prime number.")

# Part 2: Sum all primes between 2 and N (inclusive)
prime_sum = 0
for num in range(2, n + 1):
    is_prime = True
    for d in range(2, int(num ** 0.5) + 1):
        if num % d == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += num

print(f"Sum of all primes between 2 and {n}: {prime_sum}")
````

**Q35**. Take two positive integers. Compute their HCF (Highest Common Factor) using Euclidean algorithm with a `while` loop. Then compute LCM using the relation: HCF × LCM = a × b. Do not use math.gcd.

````python
a = int(input("Enter first positive integer: "))
b = int(input("Enter second positive integer: "))

x = a
y = b

while y != 0:
    remainder = x % y
    x = y
    y = remainder

hcf = x
lcm = (a * b) // hcf

print("HCF:", hcf)
print("LCM:", lcm)
````

**Q38**. Word Frequency Analyzer for Swiggy reviews. Take a review sentence. Find the frequency of each word (case-insensitive, "Food" and "food" should be counted together). Print words sorted by frequency in descending order. Also print the MOST frequent word and its count. Use a dict.

````python
review = input("Enter a Swiggy review: ").lower()
words = review.split()

freq = {}
for word in words:
    # Remove common punctuation attached to words
    clean_word = word.strip(".,!?\"'")
    if clean_word != "":
        if clean_word in freq:
            freq[clean_word] += 1
        else:
            freq[clean_word] = 1

# Sort dictionary items by count descending using simple bubble sort (no lambda)
items = list(freq.items())
for i in range(len(items)):
    for j in range(i + 1, len(items)):
        if items[j] > items[i]:
            items[i], items[j] = items[j], items[i]

print("\nWords sorted by frequency (descending):")
for word, count in items:
    print(f"{word}: {count}")

if len(items) > 0:
    print(f"\nMost frequent word: '{items}' (count: {items})")
````

(a) Languages known by BOTH sections

 (b) Languages known ONLY by Section A

 (c) Languages known ONLY by Section B

 (d) Total unique languages across both sections

 (e) Languages exclusive to one section but not both (symmetric difference)

````python
section_a = {"Python", "Java", "C++", "SQL"}
section_b = {"Java", "JavaScript", "Python", "Go", "SQL"}

# (a) Intersection: known by BOTH sections
both = section_a & section_b

# (b) Difference: known ONLY by Section A
only_a = section_a - section_b

# (c) Difference: known ONLY by Section B
only_b = section_b - section_a

# (d) Union: total unique languages across both sections
all_unique = section_a | section_b

# (e) Symmetric Difference: exclusive to one section but not both
exclusive = section_a ^ section_b

print("(a) Known by BOTH sections:", both)
print("(b) Known ONLY by Section A:", only_a)
print("(c) Known ONLY by Section B:", only_b)
print("(d) Total unique languages:", len(all_unique), all_unique)
print("(e) Exclusive to one section:", exclusive)
````

````python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

# (a) Maximum price and day
max_price = max(prices)
max_day = days[prices.index(max_price)]

# (b) Minimum price and day
min_price = min(prices)
min_day = days[prices.index(min_price)]

# (c) Average price
avg_price = sum(prices) / len(prices)

# (d) Biggest single-day gain and (e) Total volatility
biggest_gain = 0
gain_days = ""
total_volatility = 0

for i in range(1, len(prices)):
    change = prices[i] - prices[i - 1]
    total_volatility += abs(change)

    if change > biggest_gain:
        biggest_gain = change
        gain_days = f"{days[i - 1]} to {days[i]}"

print(f"(a) Maximum price: ₹{max_price} on {max_day}")
print(f"(b) Minimum price: ₹{min_price} on {min_day}")
print(f"(c) Average price: ₹{avg_price:.2f}")
print(f"(d) Biggest single-day gain: ₹{biggest_gain} ({gain_days})")
print(f"(e) Total volatility: ₹{total_volatility}")
````

**Q45**. Caesar Cipher Encryptor. Take a message and a shift value N as input. Encrypt by shifting each letter forward by N positions in the alphabet (wrapping around, Z + 1 = A). Preserve case. Keep digits, spaces, and special characters UNCHANGED. Example: "Hello, India!" with shift 3 → "Khoor, Lqgld!"
🇮🇳

````python
message = input("Enter message: ")
shift = int(input("Enter shift value N: "))
encrypted = ""

for ch in message:
    if ch.isupper():
        # 'A' has ASCII value 65
        new_char = chr((ord(ch) - 65 + shift) % 26 + 65)
        encrypted += new_char
    elif ch.islower():
        # 'a' has ASCII value 97
        new_char = chr((ord(ch) - 97 + shift) % 26 + 97)
        encrypted += new_char
    else:
        # Keep spaces, digits, and special characters unchanged
        encrypted += ch

print("Encrypted message:", encrypted)
````

**Q46**. Write a function `calculate_gst(amount, gst_rate=18, gst_inclusive=False)`:

- If gst_inclusive=False: amount is base, add GST on top -

 If gst_inclusive=True: amount already includes GST, extract it

 - Return a tuple: (base_amount, gst_amount, total_amount)

 - All values rounded to 2 decimals

 Test cases:

 calculate_gst(1000) → (1000.00, 180.00, 1180.00)

 calculate_gst(1180, gst_inclusive=True) → (1000.00, 180.00, 1180.00)

 calculate_gst(500, gst_rate=5) → (500.00, 25.00, 525.00)

````python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if not gst_inclusive:
        base_amount = amount
        gst_amount = base_amount * (gst_rate / 100)
        total_amount = base_amount + gst_amount
    else:
        total_amount = amount
        base_amount = total_amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount

    return (round(base_amount, 2), round(gst_amount, 2), round(total_amount, 2))

# Test cases
print(calculate_gst(1000))                      # (1000.0, 180.0, 1180.0)
print(calculate_gst(1180, gst_inclusive=True))  # (1000.0, 180.0, 1180.0)
print(calculate_gst(500, gst_rate=5))           # (500.0, 25.0, 525.0)
````

**Q47**. Write a RECURSIVE function `power(base, exp)` that computes base raised to exp, WITHOUT using the `**` operator and WITHOUT using `pow()`. Handle exp = 0 (return 1) and negative exp (return 1/power(base, -exp)). Test with: power(2, 10), power(5, 0), power(3, -2).

````python
def power(base, exp):
    if exp == 0:
        return 1
    elif exp < 0:
        return 1 / power(base, -exp)
    else:
        return base * power(base, exp - 1)

# Test cases
print("power(2, 10) =", power(2, 10))
print("power(5, 0)  =", power(5, 0))
print("power(3, -2) =", power(3, -2))
````

**Q49**. Zomato Weekly Sales Analytics. Create a NumPy array of 7-day Zomato Bengaluru sales (in ₹):

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])

 # Mon Tue Wed Thu Fri Sat Sun


Without using loops, compute and print:

(a) Total weekly sales

(b) Daily average

(c) The day (Mon/Tue/...) with maximum sales

(d) Number of days with sales above the weekly average

(e) Percentage growth from Monday to Sunday

(f) An array of day-on-day % changes (length 6)

````python
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

# (a) Total weekly sales
total_sales = np.sum(sales)

# (b) Daily average
daily_avg = np.mean(sales)

# (c) Day with maximum sales
max_day = days[np.argmax(sales)]

# (d) Number of days with sales above weekly average
days_above_avg = np.sum(sales > daily_avg)

# (e) Percentage growth from Monday to Sunday
mon_to_sun_growth = ((sales[-1] - sales[0]) / sales[0]) * 100

# (f) Array of day-on-day % changes (length 6)
dod_pct_change = ((sales[1:] - sales[:-1]) / sales[:-1]) * 100

print("(a) Total weekly sales: ₹", total_sales)
print(f"(b) Daily average: ₹{daily_avg:.2f}")
print("(c) Day with maximum sales:", max_day)
print("(d) Days above weekly average:", days_above_avg)
print(f"(e) Monday to Sunday growth: {mon_to_sun_growth:.2f}%")
print("(f) Day-on-day % changes:\n", np.round(dod_pct_change, 2))
````
